Problem Statement : Univariate Regression - Predict the price of the house(y) based on size(x)


```python
x_train = np.array([1.0,2.0])
y_train = np.array([300,500])
```

Computing cost: how well the model is predicting the target price of the house

```python
def compute_cost(x_train,y_train,w,b):
  m=x_train.shape[0]
  cost_sum = 0
  for i in range(m):
    f_wb = w * x_train[i] + b
    cost = (f_wb - y_train[i]) ** 2
    cost_sum = cost_sum + cost
  total_cost = (1/2*m)*cost_sum

  return total_cost

```
